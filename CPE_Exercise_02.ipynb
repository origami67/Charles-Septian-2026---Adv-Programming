```python
import math

while 1:
    n = int(input())

    if n == 0:
        break

    target = math.ceil(math.sqrt(n))
    mid = target**2 - target + 1
    diff = abs(n - mid)

    if target % 2 == 0:
        if n > mid:
            x = target
            y = target - diff
        else:
            x = target - diff
            y = target
    else:
        if n > mid:
            x = target - diff
            y = target
        else:
            x = target
            y = target - diff

    print(f"{x} {y}")
```

Language Detection

```python
case = 1

while True:
    word = input().strip()

    if word == "#":
        break

    if word == "HELLO":
        language = "ENGLISH"
    elif word == "HOLA":
        language = "SPANISH"
    elif word == "HALLO":
        language = "GERMAN"
    elif word == "BONJOUR":
        language = "FRENCH"
    elif word == "CIAO":
        language = "ITALIAN"
    elif word == "ZDRAVSTVUJTE":
        language = "RUSSIAN"
    else:
        language = "UNKNOWN"

    print(f"Case {case}: {language}")
    case += 1
```

Square Numbers

```python
import math

while True:
    a, b = map(int, input().split())

    if a == 0 and b == 0:
        break

    start = math.ceil(math.sqrt(a))
    end = math.floor(math.sqrt(b))

    answer = end - start + 1

    if answer < 0:
        answer = 0

    print(answer)
```
