Polynomial Showdown

```python
import sys

data = sys.stdin.read().strip().split()

nums = []

for x in data:
    nums.append(int(x))

i = 0

while i < len(nums):
    c = nums[i:i + 9]
    i += 9

    if len(c) < 9:
        break

    ans = ""

    for j in range(9):
        coef = c[j]
        power = 8 - j

        if coef == 0:
            continue

        if ans == "":
            if coef < 0:
                ans += "-"
        else:
            if coef < 0:
                ans += " - "
            else:
                ans += " + "

        coef = abs(coef)

        if power == 0:
            ans += str(coef)

        elif power == 1:
            if coef == 1:
                ans += "x"
            else:
                ans += str(coef) + "x"

        else:
            if coef == 1:
                ans += "x^" + str(power)
            else:
                ans += str(coef) + "x^" + str(power)

    if ans == "":
        ans = "0"

    print(ans)
```

Haiku Review

```python
vowels = "aeiouy"

def count_syllables(text):
    count = 0
    inside_vowel = False

    for ch in text:
        if ch in vowels:
            if not inside_vowel:
                count += 1
                inside_vowel = True
        else:
            inside_vowel = False

    return count


while True:
    haiku = input()

    if haiku == "e/o/i":
        break

    parts = haiku.split("/")

    first = count_syllables(parts[0])
    second = count_syllables(parts[1])
    third = count_syllables(parts[2])

    if first != 5:
        print(1)
    elif second != 7:
        print(2)
    elif third != 5:
        print(3)
    else:
        print("Y")
```

Zapping

```python
while True:
    a, b = map(int, input().split())

    if a == -1 and b == -1:
        break

    up = abs(a - b)
    down = 100 - up

    answer = min(up, down)

    print(answer)
```

Word Amalgamation

```python
dictionary = []

while True:
    word = input()

    if word == "XXXXXX":
        break

    dictionary.append(word)

dictionary.sort()

while True:
    scrambled = input()

    if scrambled == "XXXXXX":
        break

    found = False

    sorted_scrambled = sorted(scrambled)

    for word in dictionary:
        if sorted(word) == sorted_scrambled:
            print(word)
            found = True

    if found == False:
        print("NOT A VALID WORD")

    print("******")
```
