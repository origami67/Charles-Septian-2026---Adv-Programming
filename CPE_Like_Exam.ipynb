The 3n + 1 Sequence (Single)

```python
aa = []

n = int(input())
print(n, end=" ")
while n > 1:
    if n % 2 == 0:
        n = n // 2
        aa.append(n)
    else:
        n = 3 * n + 1
        aa.append(n)

    print(n, end=" ")
```

Ant on a 5x5 Chessboard

```python
import math
n = int(input())
if n == 0:
    exit()

target = math.ceil(math.sqrt(n))
mid = target**2 - target + 1
diff = abs(n - mid)

if target % 2 == 0:
    if n > mid:
        x = target
        y = target - diff
    else:
        x = target - diff
        y = target
else:
    if n > mid:
        x = target - diff
        y = target
    else:
        x = target
        y = target - diff

print(f"({x}, {y})")
```

Perfect, Deficient, or Abundant

```python
n = int(input())
c = 0
for i in range(1, n):
    if n % i == 0:
        c = c + i

if c == n:
    print("perfect")
elif c > n:
    print("abundant")
else:
    print("deficient")
```

Simple Cipher

```python
import sys

lines = [line.strip() for line in sys.stdin if line.strip()]

i = 0
a = lines[i]
b = lines[i + 1]

count_a = [0] * 26
count_b = [0] * 26

for ch in a:
    count_a[ord(ch) - ord('A')] += 1
for ch in b:
    count_b[ord(ch) - ord('A')] += 1

count_a.sort()
count_b.sort()

if count_a == count_b:
    print("YES")
else:
    print("NO")
```
