Basic wall maze

```python
from collections import deque

dx = [0, 1, 0, -1]
dy = [-1, 0, 1, 0]
move = "NESW"

while True:
    sx, sy = map(int, input().split())

    if sx == 0 and sy == 0:
        break

    ex, ey = map(int, input().split())

    wall = [[[0] * 4 for _ in range(8)] for _ in range(8)]

    for _ in range(3):
        x1, y1, x2, y2 = map(int, input().split())

        # vertical wall
        if x1 == x2:
            x = x1

            for y in range(min(y1, y2), max(y1, y2)):
                row = y + 1

                wall[x][row][1] = 1
                wall[x + 1][row][3] = 1

        # horizontal wall
        else:
            y = y1

            for x in range(min(x1, x2), max(x1, x2)):
                col = x + 1

                wall[col][y][2] = 1
                wall[col][y + 1][0] = 1

    q = deque()
    q.append((sx, sy))

    vis = [[0] * 8 for _ in range(8)]
    vis[sx][sy] = 1

    par = [[None] * 8 for _ in range(8)]

    while q:
        x, y = q.popleft()

        if x == ex and y == ey:
            break

        for i in range(4):
            if wall[x][y][i]:
                continue

            nx = x + dx[i]
            ny = y + dy[i]

            if nx < 1 or nx > 6 or ny < 1 or ny > 6:
                continue

            if vis[nx][ny]:
                continue

            vis[nx][ny] = 1
            par[nx][ny] = (x, y, move[i])
            q.append((nx, ny))

    ans = []
    x, y = ex, ey

    while (x, y) != (sx, sy):
        px, py, m = par[x][y]
        ans.append(m)
        x, y = px, py

    print("".join(ans[::-1]))
```

Ancient Cipher

```python
import sys

lines = [line.strip() for line in sys.stdin if line.strip()]

for i in range(0, len(lines), 2):
    a = lines[i]
    b = lines[i + 1]

    count_a = [0] * 26
    count_b = [0] * 26

    for ch in a:
        count_a[ord(ch) - ord('A')] += 1

    for ch in b:
        count_b[ord(ch) - ord('A')] += 1

    count_a.sort()
    count_b.sort()

    if count_a == count_b:
        print("YES")
    else:
        print("NO")
```

Checkers

```python
MOD = 1000007

t = int(input())

for case in range(1, t + 1):

    n = int(input())

    board = []

    for i in range(n):

        row = input()
        board.append(row)

        if "W" in row:
            start_y = i
            start_x = row.index("W")

    dp = [[0] * n for _ in range(n)]

    dp[start_y][start_x] = 1

    for y in range(start_y, 0, -1):

        for x in range(n):

            if dp[y][x] == 0:
                continue

            new_x = x - 1
            new_y = y - 1

            if 0 <= new_x < n and 0 <= new_y < n:

                if board[new_y][new_x] == ".":

                    dp[new_y][new_x] += dp[y][x]

                elif board[new_y][new_x] == "B":

                    jump_x = x - 2
                    jump_y = y - 2

                    if 0 <= jump_x < n and 0 <= jump_y < n:

                        if board[jump_y][jump_x] == ".":

                            dp[jump_y][jump_x] += dp[y][x]

            new_x = x + 1
            new_y = y - 1

            if 0 <= new_x < n and 0 <= new_y < n:

                if board[new_y][new_x] == ".":

                    dp[new_y][new_x] += dp[y][x]

                elif board[new_y][new_x] == "B":

                    jump_x = x + 2
                    jump_y = y - 2

                    if 0 <= jump_x < n and 0 <= jump_y < n:

                        if board[jump_y][jump_x] == ".":

                            dp[jump_y][jump_x] += dp[y][x]

    answer = sum(dp[0]) % MOD

    print(f"Case {case}: {answer}")
```

Finding Haplotypes

```python
import sys

def make_pairs(g):
    pairs = [("", "")]

    for ch in g:
        new_pairs = []

        for a, b in pairs:
            if ch == "0":
                new_pairs.append((a + "0", b + "0"))
            elif ch == "1":
                new_pairs.append((a + "1", b + "1"))
            else:
                new_pairs.append((a + "0", b + "1"))
                new_pairs.append((a + "1", b + "0"))

        pairs = new_pairs

    return pairs


def solve_case(genotypes):
    all_pairs = []

    for g in genotypes:
        all_pairs.append(make_pairs(g))

    answer = [999999]

    def dfs(index, used):
        if len(used) >= answer[0]:
            return

        if index == len(all_pairs):
            answer[0] = len(used)
            return

        for a, b in all_pairs[index]:
            old_used = used.copy()

            used.add(a)
            used.add(b)

            dfs(index + 1, used)

            used.clear()
            used.update(old_used)

    dfs(0, set())
    return answer[0]


data = sys.stdin.read().split()

i = 0
results = []

while i + 1 < len(data):
    n = int(data[i])
    l = int(data[i + 1])
    i += 2

    genotypes = data[i:i + n]
    i += n

    results.append(str(solve_case(genotypes)))

print("\n".join(results))
```
